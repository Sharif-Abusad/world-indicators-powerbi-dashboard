```python
all_dfs = []
for i in range(1, 70):  # Assuming there are 70 pages of indicators
    url = f"https://api.worldbank.org/v2/indicators?format=json&per_page=500&page={i}"
    response = requests.get(url)
    if response.status_code == 200:
        data = response.json()
        if len(data) > 1:
            indicators = data[1]
            df = pd.DataFrame([{
                "id": item["id"],
                "name": item["name"]
            } for item in indicators])
            all_dfs.append(df)
            print(f"Fetched page {i} with {len(df)} indicators.")
        else:
            break
    else:
        print(f"Failed to fetch data for page {i}, status code: {response.status_code}")
        break
```

```python
final_df = pd.concat(all_dfs, ignore_index=True)
```

```python
final_df.to_csv("indicators.csv", index=False)
```

```python
indicator_groups = {
    "economic_activity_growth": [
        "NY.GDP.MKTP.KD.ZG", # GDP growth (annual %)
        "NY.GDP.PCAP.CD"     # GDP per capita (current US$)
    ],
    "labour_market_indicators": [
        "SL.UEM.TOTL.ZS", # Unemployment total (% of total labor force) (modeled ILO estimate)
        "SL.UEM.1524.ZS", # Unemployment youth total (ages 15-24) (% of total labor force) (modeled ILO estimate)
        "SL.TLF.TOTL.IN"  # Labor force participation rate, total (% of total population ages 15+)
    ],
    "trade_globalization": [
        "NE.EXP.GNFS.CD", # Exports of goods and services (current US$) 
        "NE.IMP.GNFS.CD"  # Imports of goods and services (current US$)
    ],
    "poverty_inequality": [
        "SI.POV.NAHC", # Poverty headcount ratio at national poverty lines (% of population) 
        "SI.POV.GINI"  # GINI index (measure of income inequality)
    ],
    "environmental_indicators": [
        "EG.FEC.RNEW.ZS", # Renewable energy consumption (% of total final energy consumption)
        "AG.LND.FRST.ZS"  # Forest area (% of land area)
    ],
    "health_indicators": [
        "SP.DYN.LE00.IN",    # Life expectancy at birth (years)
        "SP.DYN.IMRT.IN",    # Infant mortality rate (per 1,000 live births)
        "SH.H2O.BASW.ZS",    # Access to atleast basic water services (% of population)
        "SH.XPD.CHEX.GD.ZS", # Current health expenditure (% of GDP)
        "SH.IMM.IDPT",       # Immunization, DPT (% of children ages 12-23 months)
        "SH.IMM.MEAS",       # Immunization, measles (% of children ages 12-23 months)
        "SH.MMR.RISK.ZS",    # Risk of maternal death (% of women ages 15-49)
        "SH.DTH.COMM.ZS",    # Deaths from communicable diseases (% of total deaths)
        "SH.TBS.INCD",       # Tuberculosis incidence (per 100,000 people)
        "SH.STA.BRTC.ZS",    # Births attended by skilled health personnel (% of total births)
        "SH.STA.MMRT",       # Maternal mortality ratio (model estimate, per 100,000 live births)
        "SP.POP.65UP.TO.ZS", # Population aged 65 and above (% of total population)
        "SH.HIV.INCD.ZS"     # HIV incidence rate (per 1,000 uninfected population ages 15-49)
    ],
    "technology_indicators": [
        "IT.NET.USER.ZS", # Individuals using the Internet (% of population)
        "IT.CEL.SETS.P2"  # Mobile cellular subscriptions (per 100 people)
    ]
}
```

```python
import time
base_url = f'https://api.worldbank.org/countries/all/indicators/{}?format=json&per_page=1000&page={}'

category_dfs = {}
for category, indicators in indicator_groups.items():
    print(f"Fetching data for category: {category}")
    all_dfs_for_category = []

    for indicator_code in indicators:
        print(f"Fetching data for indicator: {indicator_code}")
        page = 1

        while True:
            url = base_url.format(indicator_code, page)
            response = requests.get(url)

            if response.status_code != 200:
                print(f"Failed to fetch data for indicator {indicator_code}, page {page}, status code: {response.status_code}")
                break

            data = response.json()
            
            if len(data) < 2 or not data[1]:
                print(f"No more data for indicator {indicator_code} at page {page}.")
                break

            total_pages = data[0].get('pages', 1)
            records = data[1]

            df = pd.json_normalize(records)
            df = df[[
                "country.id", "country.value", "indicator.id", 
                "indicator.value", "date", "value"
            ]]
            
            df.rename(columns={
                "country.id": "country_id", 
                "country.value": "country_name", 
                "indicator.id": "indicator_id",
                "indicator.value": "indicator_name",
                "date": "year",
            }, inplace=True)

            df = df[df["year"].astype(int) > 2015]
            all_dfs_for_category.append(df)

            if page >= total_pages:
                print(f"All pages fetched for indicator {indicator_code}.")
                break
            else:
                page += 1
                time.sleep(0.5)  # Sleep for 0.5 seconds to avoid hitting the API too quickly

    if all_dfs_for_category:
        combined_df = pd.concat(all_dfs_for_category, ignore_index=True)
        category_dfs[category] = combined_df
        print(f"Total rows fetched for category {category}: {len(combined_df)}")
    else:
        print(f"No data fetched for category {category}.")

print("Data fetching completed for all categories.")
```
